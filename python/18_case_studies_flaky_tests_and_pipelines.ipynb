## Part 1: The 50-Sided Die Trap & Why 0/100 Passes Proves Almost Nothing

### The Scenario
A test fails twice in 100 CI runs on the main branch ($p = 2.0\%$). An engineer
submits a bug fix, executes the test 100 times in a local loop, observes zero
failures ($0/100$), declares the issue resolved, and merges. Two days later,
the test flakes again in release qualification.

### The Mathematics of the 50-Sided Die
A $2\%$ flake rate represents a fair **50-sided die** with 1 red face. The
probability of observing zero red faces in $N = 100$ independent rolls is:
$$P(0 \text{ failures} \mid p = 0.02, N = 100) = (1 - 0.02)^{100}
= 0.98^{100} \approx \mathbf{13.3\%}$$

A $13.3\%$ probability is almost as likely as rolling a 6 on a standard
six-sided gaming die ($16.7\%$). Observing 100 clean passes is weak
circumstantial evidence, not definitive proof of a fix.

### The Frequentist Rule of Three vs. Bayesian Posterior
Under classical statistics, observing zero events in $N$ trials yields a
$95\%$ upper confidence bound governed by the **Rule of Three**:
$$p_{\text{upper}} \approx \frac{3}{N} = \frac{3}{100} = \mathbf{3.0\%}$$
The test began with a $2.0\%$ flake rate; after 100 clean passes, the
frequentist
guaranteed upper bound is $3.0\%$!

Under Bayesian updating with a flat prior $p \sim \text{Beta}(1, 1)$, observing
zero failures in $N$ trials yields the posterior:
$$p \mid D \sim \text{Beta}(1, N + 1)$$
The exact posterior probability that the flake rate exceeds $1\%$ is:
$$P(p > 0.01 \mid D) = (1 - 0.01)^{101} = 0.99^{101} \approx \mathbf{36.2\%}$$
There remains a $36.2\%$ chance that the test still violates a $1\%$ SLA!

````python
# ==============================================================================
# Part 1: The 50-Sided Die Simulation & Posterior Flake Rate Distribution
# ==============================================================================
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

n_runs = np.arange(1, 201)
flake_rates = [0.005, 0.01, 0.02, 0.05]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 1. Probability of zero failures across trial counts
for p in flake_rates:
  p_zero = (1.0 - p) ** n_runs
  ax1.plot(n_runs, p_zero, label=f"True Flake Rate p = {p * 100:.1f}%")

p_zero_100 = (1.0 - 0.02) ** 100
ax1.plot(
    100,
    p_zero_100,
    "ro",
    markersize=8,
    label=f"N=100, p=2%: P(0 fail) = {p_zero_100 * 100:.1f}%",
)
ax1.axvline(100, color="gray", linestyle="--", alpha=0.5)
ax1.axhline(0.05, color="red", linestyle=":", alpha=0.6, label="5% Alpha Line")
ax1.set_xlabel("Number of Consecutive Test Runs (N)")
ax1.set_ylabel("Probability of Zero Failures P(0 in N)")
ax1.set_title(
    "Probability of Zero Failures (1 - p)^N\n(The 50-Sided Die Trap)"
)
ax1.grid(True, alpha=0.3)
ax1.legend(loc="upper right", fontsize=9)

# 2. Beta-Binomial posterior densities after 0 failures in N runs
theta = np.linspace(0.0, 0.10, 500)
sample_sizes = [20, 50, 100, 200]
colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728"]

for n, col in zip(sample_sizes, colors):
  posterior = stats.beta.pdf(theta, 1, n + 1)
  q95 = stats.beta.ppf(0.95, 1, n + 1)
  rule_of_three = 3.0 / n
  ax2.plot(
      theta * 100,
      posterior,
      label=f"N = {n} (95% CI < {q95 * 100:.1f}%, Rule3 = {rule_of_three * 100:.1f}%)",
      color=col,
      linewidth=2,
  )
  ax2.axvline(q95 * 100, color=col, linestyle="--", alpha=0.7)

ax2.set_xlabel("Flake Rate θ (%)")
ax2.set_ylabel("Posterior Probability Density p(θ | D)")
ax2.set_title("Posterior Belief After 0 Failures in N Runs\n(Beta(1, N + 1))")
ax2.grid(True, alpha=0.3)
ax2.legend(loc="upper right", fontsize=9)
plt.tight_layout()
plt.show()

q95_100 = stats.beta.ppf(0.95, 1, 101)
rule_three_100 = 3.0 / 100.0
bf_point = 1.0 / ((1.0 - 0.02) ** 100)
print(f"Results for N = 100 clean passes with historical flake rate p = 2.0%:")
print(f"  P(0 failures | p = 2.0%, N = 100)  = {p_zero_100 * 100:.2f}%")
print(f"  Bayesian 95% Upper Bound on Rate   = {q95_100 * 100:.2f}%")
print(f"  Frequentist Rule of Three (3 / N)  = {rule_three_100 * 100:.2f}%")
print(f"  Likelihood Ratio H_fix / H_unfixed = {bf_point:.2f}x")
````

## Part 2: The (N, M) Passing Streak Trap & Gambler's Fallacy in CI Gatekeeping

### The Promotion Dilemma
Engineering platforms commonly use heuristic streak criteria to promote test
targets from non-blocking Staging suites to release-blocking presubmit gates:
> *"Promote test $T$ to blocking tier if it has achieved a historical passing
> streak of at least $N = 50$ consecutive runs, and its current streak is at
> least $M = 10$."*

### Why Is This Heuristic Defective?
1. **Survivorship Bias & The Gambler's Fallacy**:
   - For a test with a permanent $p = 2.0\%$ flake rate, the probability of
     passing 50 consecutive runs in any single attempt is $0.98^{50} \approx
     36.4\%$.
   - Across $500$ runs in staging, the probability that the test experiences at
     least one 50-run passing streak purely by chance exceeds **$65\%$**!
   - Streak counters reward lucky clusters of passes while ignoring the
     underlying Poisson failure rate.
2. **Amnesia**:
   - A streak counter looks solely at recent runs and entirely forgets that the
     test failed 4 times immediately prior to the streak.

### The Bayesian Alternative: Continuous Dynamic Memory Filter
Instead of brittle streak counters, Chapter 15 demonstrated how an exponential
memory decay filter ($\\gamma = 0.98$) continuously tracks state:
$$\alpha_t = \gamma \alpha_{t-1} + (1 - y_t), \quad
\beta_t = \gamma \beta_{t-1} + y_t$$

A test is promoted if and only if:
$$P(\theta < 1.0\% \mid \alpha_t, \beta_t) \ge \mathbf{95\%}$$
This rule requires sustained health, cannot be fooled by lucky streaks, and
automatically demotes deteriorating tests without manual intervention.

````python
# ==============================================================================
# Part 2: The (N, M) Passing Streak Trap vs. Bayesian Dynamic Memory Filter
# ==============================================================================
np.random.seed(42)

n_total_runs = 500
true_flake_rate = 0.02  # 2.0% flake rate
gamma_run = 0.98 ** (1.0 / 20.0)  # calibrated for ~20 runs per day
target_sla = 0.01  # 1.0% maximum allowable rate for promotion

outcomes = np.random.binomial(1, true_flake_rate, size=n_total_runs)

# 1. Track heuristic streak: N = 50 historical, M = 10 current
current_streak = np.zeros(n_total_runs, dtype=int)
max_streak_so_far = np.zeros(n_total_runs, dtype=int)
promoted_streak = np.zeros(n_total_runs, dtype=bool)

streak = 0
max_streak = 0
for t in range(n_total_runs):
  if outcomes[t] == 0:
    streak += 1
  else:
    streak = 0
  current_streak[t] = streak
  max_streak = max(max_streak, streak)
  max_streak_so_far[t] = max_streak
  if max_streak >= 50 and streak >= 10:
    promoted_streak[t] = True

# 2. Track Bayesian dynamic discount filter
alpha_hist = np.zeros(n_total_runs)
beta_hist = np.zeros(n_total_runs)
prob_healthy = np.zeros(n_total_runs)
posterior_mean = np.zeros(n_total_runs)

alpha = 1.0
beta = 1.0

for t in range(n_total_runs):
  alpha = alpha * gamma_run + (1 - outcomes[t])
  beta = beta * gamma_run + outcomes[t]
  alpha_hist[t] = alpha
  beta_hist[t] = beta
  posterior_mean[t] = beta / (alpha + beta)
  prob_healthy[t] = stats.beta.cdf(target_sla, beta, alpha)

promoted_bayes = prob_healthy >= 0.95

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 8), sharex=True)

# Panel 1: Heuristic Streak Trajectory
ax1.plot(
    range(n_total_runs),
    current_streak,
    label="Current Passing Streak",
    color="#1f77b4",
    alpha=0.7,
)
ax1.plot(
    range(n_total_runs),
    max_streak_so_far,
    label="Max Historical Streak",
    color="#2ca02c",
    linewidth=2,
)
ax1.axhline(50, color="red", linestyle="--", label="Streak Threshold (N = 50)")
ax1.fill_between(
    range(n_total_runs),
    0,
    1,
    where=promoted_streak,
    color="orange",
    alpha=0.25,
    transform=ax1.get_xaxis_transform(),
    label="Streak Heuristic: FAULTILY PROMOTED",
)
ax1.set_ylabel("Passing Streak (Runs)")
ax1.set_title(
    "The (N, M) Passing Streak Trap: Flaky Test (p=2%) Sneaking Through Gates"
)
ax1.grid(True, alpha=0.3)
ax1.legend(loc="upper left")

# Panel 2: Bayesian Dynamic Memory Filter
ax2.plot(
    range(n_total_runs),
    posterior_mean * 100,
    label="Posterior Mean Flake Rate (%)",
    color="#9467bd",
    linewidth=2,
)
ax2.axhline(
    target_sla * 100,
    color="black",
    linestyle=":",
    label=f"Target SLA ({target_sla * 100:.1f}%)",
)
ax2.plot(
    range(n_total_runs),
    prob_healthy * 100,
    label="Bayesian Confidence P(θ < 1.0%)",
    color="#2ca02c",
    linewidth=1.5,
)
ax2.axhline(
    95,
    color="green",
    linestyle="--",
    label="Bayesian Promotion Gate (95% Confidence)",
)
ax2.set_xlabel("Test Run Index (Timeline)")
ax2.set_ylabel("Probability / Rate (%)")
ax2.set_title("Bayesian Dynamic Memory Filter: Robustly Blocking Promotion")
ax2.grid(True, alpha=0.3)
ax2.legend(loc="upper right")

plt.tight_layout()
plt.show()

print(f"Summary over {n_total_runs} CI runs (True Flake Rate p = {true_flake_rate * 100:.1f}%):")
print(f"  Maximum streak achieved by chance: {max(max_streak_so_far)} passes")
print(f"  Runs falsely approved by streak heuristic: {int(np.sum(promoted_streak))}")
print(f"  Runs permitted by Bayesian 95% gate:       {int(np.sum(promoted_bayes))}")
````

## Part 3: End-to-End Reliability Engineering Decision Router

How do modern reliability platforms govern testing pipelines? Rather than
relying on ad-hoc rules, they synthesize the entire Bayesian curriculum into
an automated **Four-Pillar Decision Router**:

```text
               END-TO-END BAYESIAN RELIABILITY GOVERNANCE PIPELINE
               
  [Incoming Patch] 
         │
         ▼
  ┌────────────────────────────────────────────────────────────────────────┐
  │ 1. Fix Verification (Chapter 19 / SPRT)                                 │
  │    Sequential Bayes early-abort: Pass -> -0.019 pts, Fail -> +3.00 pts   │
  │    Saves >90% compute when verifying race condition fixes               │
  └───────────────────────────────────┬────────────────────────────────────┘
                                      │ Pass
                                      ▼
  ┌────────────────────────────────────────────────────────────────────────┐
  │ 2. Test Promotion & Health Meter (Chapter 15 / Chapter 20)                 │
  │    Exponential memory decay (gamma = 0.98) tracking alpha and beta      │
  │    Gate: P(Flake Rate < SLA) >= 95% (Immune to lucky streak traps)      │
  └───────────────────────────────────┬────────────────────────────────────┘
                                      │ Promoted to Suite
                                      ▼
  ┌────────────────────────────────────────────────────────────────────────┐
  │ 3. Suite Tolerance Contracts (Chapters 17 & 21)                        │
  │    Serial Multiplier: P(All Pass) = (1 - p)^K                           │
  │    Allocates SRE error budgets across K tests to prevent suite collapse │
  └───────────────────────────────────┬────────────────────────────────────┘
                                      │ Regression Detected
                                      ▼
  ┌────────────────────────────────────────────────────────────────────────┐
  │ 4. Active Culprit Localization (Chapter 22)                            │
  │    Marginalize unknown rates via Beta conjugacy (O(N) prefix sweep)     │
  │    Active test dispatch via Mutual Information I(Trial; Culprit)        │
  └────────────────────────────────────────────────────────────────────────┘
```

Below, we simulate this complete governance loop and render an integrated
four-panel summary dashboard.

````python
# ==============================================================================
# Part 3: End-to-End Reliability Engineering Decision Router Dashboard
# ==============================================================================
fig, axes = plt.subplots(2, 2, figsize=(15, 11))

# --- Panel 1: Stage 1 Fix Verification (Wald's SPRT Early-Abort) ---
p_broken = 0.05
p_fixed = 0.001
log_pass = np.log((1 - p_broken) / (1 - p_fixed))  # negative increment
log_fail = np.log(p_broken / p_fixed)              # positive sledgehammer

b_abort = np.log(19.0)   # upper abort threshold (+2.94)
a_accept = np.log(1/19.0) # lower accept threshold (-2.94)

# Stream: 2 passes, then a failure on run 3
sprt_stream = [0, 0, 1]
llr_trace = [0.0]
for outcome in sprt_stream:
  inc = log_fail if outcome == 1 else log_pass
  llr_trace.append(llr_trace[-1] + inc)

axes[0, 0].step(range(len(llr_trace)), llr_trace, where="post", color="#d62728", lw=2, label="LLR Trace")
axes[0, 0].axhline(b_abort, color="red", linestyle="--", label="Abort Boundary (+2.94)")
axes[0, 0].axhline(a_accept, color="green", linestyle="--", label="Accept Boundary (-2.94)")
axes[0, 0].set_title("Stage 1: Fix Verification (SPRT Early-Abort)")
axes[0, 0].set_xlabel("Run Count")
axes[0, 0].set_ylabel("Log-Likelihood Ratio")
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].legend(loc="lower left", fontsize=8)

# --- Panel 2: Stage 2 Test Promotion Confidence for 4 Test Profiles ---
sample_n = np.arange(10, 301, 10)
profiles = [("Rock-Solid (0.1%)", 0.001, "#2ca02c"),
            ("Borderline (0.9%)", 0.009, "#1f77b4"),
            ("Flaky (2.0%)", 0.020, "#ff7f0e"),
            ("Severe (4.0%)", 0.040, "#d62728")]

for label, p, col in profiles:
  conf = [stats.beta.cdf(0.01, 1 + n * p, 1 + n * (1 - p)) * 100 for n in sample_n]
  axes[0, 1].plot(sample_n, conf, label=label, color=col, lw=2)

axes[0, 1].axhline(95, color="black", linestyle="--", label="Promotion Gate (95%)")
axes[0, 1].set_title("Stage 2: Test Promotion Confidence P(θ < 1.0%)")
axes[0, 1].set_xlabel("Effective Sample Size (N_eff)")
axes[0, 1].set_ylabel("Confidence (%)")
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].legend(loc="center right", fontsize=8)

# --- Panel 3: Stage 3 Suite Tolerance Contracts (The Serial Multiplier) ---
k_suite = np.linspace(10, 2000, 100)
test_rates = [0.0001, 0.0005, 0.0010, 0.0020]
for p in test_rates:
  p_suite = ((1.0 - p) ** k_suite) * 100
  axes[1, 0].plot(k_suite, p_suite, label=f"Per-test p = {p * 100:.2f}%", lw=2)

axes[1, 0].axhline(99, color="green", linestyle=":", label="99% SRE Budget")
axes[1, 0].set_title("Stage 3: Suite Availability P(All Pass) = (1 - p)^K")
axes[1, 0].set_xlabel("Suite Size (K Tests)")
axes[1, 0].set_ylabel("Clean Suite Run (%)")
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].legend(loc="upper right", fontsize=8)

# --- Panel 4: Stage 4 Active Culprit Localization (Posterior over Commits) ---
commits = ["c0", "c1", "c2", "c3"]
post_probs = [0.055, 0.095, 0.652, 0.199]
bars = axes[1, 1].bar(commits, [p * 100 for p in post_probs], color=["#aec7e8", "#aec7e8", "#1f77b4", "#aec7e8"])
axes[1, 1].set_title("Stage 4: Culprit Localization Posterior P(C = c | D)")
axes[1, 1].set_xlabel("Candidate Revisions")
axes[1, 1].set_ylabel("Posterior Probability (%)")
axes[1, 1].grid(True, alpha=0.3, axis="y")
for bar in bars:
  yval = bar.get_height()
  axes[1, 1].text(bar.get_x() + bar.get_width()/2.0, yval + 1.5, f"{yval:.1f}%", ha="center", fontsize=9)

plt.tight_layout()
plt.show()
````

### 🔴 Tier 3 (Production Challenge): Asymmetric Early-Abort Thresholding
> **Scenario**: In a high-throughput pipeline:
> - Test execution cost: $C_{\text{run}} = \$0.50$ per run.
> - Outage cost from releasing a flaky bug: $C_{\text{FN}} = \$20{,}000$.
> - Wasted triage cost from falsely rejecting a good fix: $C_{\text{FP}} =
\$500$.
>
> 1. Derive the calibrated error probabilities $\alpha =
\frac{C_{\text{run}}}{C_{\text{FP}}}$
