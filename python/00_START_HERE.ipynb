## 🗺️ The Big Picture: The Quest to Conquer the Denominator

Every single concept in this course exists to solve **one central mathematical bottleneck**.

In Bayes' Theorem, our goal is to compute the posterior probability distribution of our parameters $\theta$ given observed data $y$:

$$\mathbf{P(\theta \mid y) = \frac{\mathcal{L}(y \mid \theta) \cdot P(\theta)}{P(y)} = \frac{\mathcal{L}(y \mid \theta) \cdot P(\theta)}{\int \mathcal{L}(y \mid \theta) \cdot P(\theta) \, d\theta}}$$

* **The Numerator is Easy**: $\mathcal{L}(y \mid \theta) \cdot P(\theta)$ is simply the likelihood of the data multiplied by our prior belief. We can easily calculate this value for any specific candidate $\theta$.
* **The Denominator is the "Boss Battle"**: $P(y) = \int \mathcal{L}(y \mid \theta) P(\theta) \, d\theta$ (the *marginal likelihood* or *evidence*) represents the **total volume under the unnormalized posterior curve across the entire universe of parameter space**.

In 1 parameter, calculating this volume is manageable with numerical integration. But in realistic scientific or production models with dozens or hundreds of parameters, computing this high-dimensional integral is **physically impossible**—it would take all the supercomputers on Earth longer than the age of the universe.

### The 4 Escalating Strategies
Every sheet in this curriculum represents a distinct computational strategy to either solve, approximate, or completely bypass that denominator:

## 🧭 The Practitioner's Strategy Matrix

When facing a real-world Bayesian problem, how do you know which tool to choose?

| Strategy | Methods | Typical Dimensions ($D$) | Execution Speed | Primary Strengths | Fatal Failure Mode | Sheet |
| :--- | :--- | :--- | :--- | :--- | :--- | :--- |
| **1. Analytical** | Beta-Binomial, Normal-Normal Conjugacy | $D = 1$ | Instantaneous (closed-form) | Exact closed-form math; ideal for high-throughput streaming updates. | **Rigid math**: Fails on non-standard priors, complex likelihoods, or latent variables. | **01, 01b** |
| **2. Brute-Force** | Grid Approximation | $D \le 2$ | Fast for 2D (vectorized) | Exact posterior shape without analytical integration; handles multimodal/skewed shapes. | **Curse of Dimensionality**: Grid points scale as $G^D$. In $D=10$, $100^{10} = 10^{20}$ points. Dead in $D \ge 3$. | **02** |
| **3. Parabolic** | Laplace / Quadratic Approximation (QUAP) | Up to $D \approx 50$ | Very Fast (milliseconds) | Fast gradient optimization to the mode (MAP); inverts Hessian ($H^{-1}$) for covariance. | **Gaussian Blindness**: Assumes symmetric bell curve. Fails on parameter boundaries ($\sigma > 0$), heavy tails, skew, multimodality. | **03** |
| **4. MCMC Engine** | Markov Chains & Detailed Balance | Arbitrary | Foundational Math | Proves that an unnormalized pairwise ratio $\frac{\pi(B)}{\pi(A)}$ bypasses $P(y)$ entirely. | Purely discrete conceptual foundation; requires translation to continuous space. | **03b** |
| **5. Continuous MCMC** | Metropolis-Hastings from Scratch | Up to $D \approx 20$ | Seconds | Universal sampler; visits states in exact proportion to posterior without knowing $P(y)$. | **Random Walk Diffusion**: In high dimensions ($D > 50$), diffusive random walks get lost in empty volume (Curse of Volume). | **04** |
| **6. Production MCMC** | Multi-Chain Stan HMC & Diagnostics | Hundreds to Thousands | Minutes | Hamiltonian physics (gradients) glides across high-dimensional geometries; $\hat{R}$ & $ESS$ verify convergence. | Requires differentiable likelihoods; high computational overhead per leapfrog step. | **05** |
| **7. Dynamic Filtering** | Exponential Bayesian Memory Decay | Any | Fast Online | Adapts to non-stationary environments and concept drift by discounting stale data. | Over-forgetting leads to high posterior variance; under-forgetting creates lag behind regime shifts. | **06** |
| **8. Decision Science** | Bayesian Decision Theory & Loss Functions | Any | Instant Post-Processing | Converts full posterior uncertainty into optimal business actions under asymmetric risk. | Defaulting to the posterior mean when loss is asymmetric (e.g., underestimating capacity) destroys business value. | **07** |
