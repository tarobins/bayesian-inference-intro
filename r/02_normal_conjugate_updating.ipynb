## 🎯 The Core Intuition: The Bayesian "Tug-of-War"

When you update your beliefs using Bayes' Rule, the resulting **Posterior Distribution** represents an optimal compromise between two competing sources:
1. **Your Prior State of Knowledge**: What you believed before collecting data ($p(\mu)$).
2. **Your Observed Evidence / Likelihood**: What your sensors, instruments, or measurements report ($p(y \mid \mu)$).

> **How does Bayes' Rule decide who gets more say in the compromise?**  
> Under a Gaussian (Normal) model, the answer is fundamental to all of data science:  
> $$\textbf{Voting power is inversely proportional to noise (variance).}$$

The source with **smaller noise (higher certainty)** gets the lion's share of the vote, while the source with **larger noise** is discounted.

### The Limits of Discrete Counts: Why We Need Continuous Conjugacy
In Chapter 1, the Beta-Binomial model turned discrete pass/fail tallies into simple pebble addition. However, physical and engineering systems rarely produce tidy binary counts—they produce **continuous measurements along the real number line**:
* **Network packet latency** in milliseconds (e.g., $142.3\text{ ms}$, $185.7\text{ ms}$).
* **Processor core temperatures** in degrees Celsius (e.g., $68.4^\circ\text{C}$, $74.1^\circ\text{C}$).
* **Autonomous vehicle distances** to an obstacle measured by radar or sonar (e.g., $10.2\text{ m}$, $13.8\text{ m}$).

Here, the parameter we wish to estimate is not a bounded probability $p \in [0, 1]$, but an unbounded continuous mean $\mu \in (-\infty, \infty)$.

### The Mechanical Tug-of-War: Springs on a Frictionless Slider
To see how Bayes' Rule combines a continuous Gaussian prior with Gaussian data, imagine a frictionless slider on a rail whose position represents our estimate of $\mu$. Attach two mechanical springs to opposite sides:
* **Spring 0 (The Prior)** is anchored at $\mu_0$ with spring stiffness $k_0 = \tau_0 = 1/\sigma_0^2$.
* **Spring L (The Likelihood)** is anchored at $y$ with spring stiffness $k_L = \tau_{\text{data}} = 1/\sigma^2$.

```text
                     THE SPRING-LOADED GAUSSIAN TUG-OF-WAR

       Prior Anchor                                       Data Measurement
      (Position mu_0)                                       (Position y)
            │                                                    │
            ▼                                                    ▼
          [mu_0] ======/\/\/\/\/\====== [mu_post] ======/\/\/\/\/\====== [y]
                     Spring 0 (tau_0)    Slider       Spring L (tau_data)
                       (Stiffness)     Equilibrium        (Stiffness)

    Net Force Balance:  tau_0 * (mu_post - mu_0) = tau_data * (y - mu_post)
    Total Stiffness:    tau_post = tau_0 + tau_data  (Precisions Simply Add!)
```

At mechanical equilibrium, the Hooke's Law restoring forces cancel ($F_0 + F_L = 0$):
$$\tau_0 (\mu_{\text{post}} - \mu_0) + \tau_{\text{data}} (\mu_{\text{post}} - y) = 0 \implies \mu_{\text{post}} = \frac{\tau_0 \mu_0 + \tau_{\text{data}} y}{\tau_0 + \tau_{\text{data}}}$$
Because the two springs pull **in parallel**, their stiffnesses add ($\tau_{\text{post}} = \tau_0 + \tau_{\text{data}}$), making the combined system stiffer (and the posterior variance $\sigma_{\text{post}}^2 = 1/\tau_{\text{post}}$ strictly smaller) than either source alone.

## Part 1: Anatomy of the Weighting Formulas

Let:
* $\mu_0, \sigma_0$: Prior mean and prior standard deviation (prior variance $\sigma_0^2$).
* $y, \sigma$: Observed data measurement and instrument noise standard deviation (measurement variance $\sigma^2$).

### 1.1 What are $w_0$ and $w_L$?
The letter **$w$ stands for "weight"** (influence):
* **$w_0 = \frac{1}{\sigma_0^2}$**: The weight of the **Prior** (subscript $0$ denotes the baseline prior).
* **$w_L = \frac{1}{\sigma^2}$**: The weight of the **Likelihood / Data** ($L$ stands for Likelihood).

*(In classical statistics textbooks, inverse variance $\tau = \frac{1}{\sigma^2}$ is called **precision**. Modern tools like PyMC and Stan parameterize with standard deviation $\sigma$, but the concept is identical: smaller error $\implies$ larger weight).*

### 1.2 What is the $W_{\text{data}}$ Formula?
Just like calculating a student's course grade from Homework ($30\%$) and an Exam ($70\%$):
$$\text{Exam Share} = \frac{\text{Exam Weight}}{\text{Total Weight}}$$

In Bayesian updating, the **total weight** is $w_0 + w_L$. The percentage share of the vote awarded to each source is:
$$W_{\text{data}} = \frac{w_L}{w_0 + w_L} = \frac{\text{Data Weight}}{\text{Total Weight}}$$
$$W_{\text{prior}} = \frac{w_0}{w_0 + w_L} = \frac{\text{Prior Weight}}{\text{Total Weight}}$$

### 1.3 The Posterior Mean Formula: $\mu_{\text{post}}$
The posterior mean is the **weighted average** of the prior mean and the observed data:
$$\mu_{\text{post}} = \frac{w_0 \mu_0 + w_L y}{w_0 + w_L} = W_{\text{prior}} \cdot \mu_0 + W_{\text{data}} \cdot y$$

And the **posterior variance** $\sigma_{\text{post}}^2$ is simply the inverse of the total weight:
$$\sigma_{\text{post}}^2 = \frac{1}{w_0 + w_L} = \frac{1}{\frac{1}{\sigma_0^2} + \frac{1}{\sigma^2}}$$
*(Because $w_0 + w_L > w_0$ and $w_0 + w_L > w_L$, the posterior variance is **always strictly smaller** than both the prior and data variances! Every piece of data reduces uncertainty).*

````r
# Part 1 Demo: The Spring-Loaded Tug-of-War (Equilibrium = Posterior Mode)
mu_0 <- 10.0; sigma_0 <- 2.0  # Prior anchor & noise (ultrasonic/dead-reckoning)
y <- 14.0;    sigma <- 1.0    # Data anchor & noise (optical LIDAR)

tau_0 <- 1.0 / (sigma_0^2)
tau_data <- 1.0 / (sigma^2)
tau_post <- tau_0 + tau_data
mu_post <- (tau_0 * mu_0 + tau_data * y) / tau_post
sigma_post <- sqrt(1.0 / tau_post)

cat(sprintf("Spring 0 stiffness (tau_0):     %.4f\n", tau_0))
cat(sprintf("Spring L stiffness (tau_data):  %.4f\n", tau_data))
cat(sprintf("Total stiffness (tau_post):     %.4f  (sigma_post = %.4f)\n", tau_post, sigma_post))
cat(sprintf("Equilibrium position (mu_post): %.4f\n", mu_post))

mu_grid <- seq(5.0, 18.0, length.out = 500)
potential_energy <- 0.5 * tau_0 * (mu_grid - mu_0)^2 + 0.5 * tau_data * (mu_grid - y)^2

par(mfrow = c(1, 2), mar = c(4.5, 4.5, 3.0, 1.0))

plot(mu_grid, potential_energy, type = "l", col = "purple", lwd = 2.5,
     main = "Mechanical Analogy: Minimum Potential Energy",
     xlab = "Position / Parameter mu", ylab = "Spring Potential Energy U(mu)")
abline(v = mu_0, col = "steelblue", lty = 2, lwd = 1.5)
abline(v = y, col = "darkorange", lty = 2, lwd = 1.5)
abline(v = mu_post, col = "firebrick", lwd = 2)
legend("top", legend = c("Total U(mu)", sprintf("Prior mu_0=%.1f", mu_0),
       sprintf("Data y=%.1f", y), sprintf("Equilibrium=%.2f", mu_post)),
       col = c("purple", "steelblue", "darkorange", "firebrick"),
       lty = c(1, 2, 2, 1), lwd = c(2.5, 1.5, 1.5, 2), cex = 0.75, bty = "n")

d_prior <- dnorm(mu_grid, mu_0, sigma_0)
d_like  <- dnorm(mu_grid, y, sigma)
d_post  <- dnorm(mu_grid, mu_post, sigma_post)
plot(mu_grid, d_post, type = "l", col = "firebrick", lwd = 2.5,
     ylim = c(0, max(d_post, d_like) * 1.15),
     main = "Statistical Equivalent: Normal-Normal Posterior",
     xlab = "Parameter mu", ylab = "Density")
lines(mu_grid, d_prior, col = "steelblue", lty = 2, lwd = 2)
lines(mu_grid, d_like, col = "darkorange", lty = 4, lwd = 2)
legend("topright", legend = c("Prior p(mu)", "Likelihood p(y|mu)", "Posterior p(mu|y)"),
       col = c("steelblue", "darkorange", "firebrick"),
       lty = c(2, 4, 1), lwd = c(2, 2, 2.5), cex = 0.75, bty = "n")
````

## Part 2: First-Principles Mathematical Derivation

Why does multiplying two Gaussian curves produce this exact weighted average?  
Let us derive it step-by-step from **Bayes' Theorem**:

$$p(\mu \mid y) \propto p(\mu) \times p(y \mid \mu)$$

### Step A: Write the Probability Density Functions
A standard Normal curve centered at mean $m$ with variance $v$ has density:
$$p(x) = \frac{1}{\sqrt{2\pi v}} \exp\left( -\frac{(x - m)^2}{2v} \right) \propto \exp\left( -\frac{1}{2} \cdot \frac{1}{v} \cdot (x - m)^2 \right)$$

Substitute weights $w_0 = \frac{1}{\sigma_0^2}$ and $w_L = \frac{1}{\sigma^2}$:
* **Prior**: $p(\mu) \propto \exp\left( -\frac{1}{2} w_0 (\mu - \mu_0)^2 \right)$
* **Likelihood**: $p(y \mid \mu) \propto \exp\left( -\frac{1}{2} w_L (y - \mu)^2 \right) = \exp\left( -\frac{1}{2} w_L (\mu - y)^2 \right)$

### Step B: Multiply Them (Add the Exponents)
When multiplying exponential functions, their powers add:
$$\text{Exponent} = -\frac{1}{2} \Big[ w_0 (\mu - \mu_0)^2 + w_L (\mu - y)^2 \Big]$$

### Step C: Expand the Quadratic Brackets
$$\begin{aligned}
w_0 (\mu - \mu_0)^2 + w_L (\mu - y)^2 &= w_0 (\mu^2 - 2\mu_0\mu + \mu_0^2) + w_L (\mu^2 - 2y\mu + y^2) \\
&= \mathbf{(w_0 + w_L)}\mu^2 - 2\mathbf{(w_0\mu_0 + w_L y)}\mu + (w_0\mu_0^2 + w_L y^2)
\end{aligned}$$
*(The trailing constant $w_0\mu_0^2 + w_L y^2$ does not depend on $\mu$, so it is absorbed into the normalizing denominator).*

### Step D: Complete the Square
Any Gaussian density centered at an unknown mode $\mu_{\text{post}}$ with precision $C$ matches the template:
$$-\frac{1}{2} C (\mu - \mu_{\text{post}})^2 = -\frac{1}{2} \Big[ C\mu^2 - 2(C \cdot \mu_{\text{post}})\mu + C\mu_{\text{post}}^2 \Big]$$

Matching coefficients:
1. **Coefficient of $\mu^2$**:
   $$C = \mathbf{w_0 + w_L} = \frac{1}{\sigma_0^2} + \frac{1}{\sigma^2} \implies \sigma_{\text{post}}^2 = \frac{1}{C} = \mathbf{\frac{1}{w_0 + w_L}}$$

2. **Coefficient of $-2\mu$**:
   $$C \cdot \mu_{\text{post}} = w_0\mu_0 + w_L y \implies \mu_{\text{post}} = \frac{w_0\mu_0 + w_L y}{C} = \mathbf{\frac{w_0 \mu_0 + w_L y}{w_0 + w_L}} \quad \blacksquare$$

## Part 3: The Three Archetypes of Bayesian Evidence (Simulation)

To build physical intuition for the inverse-variance tug-of-war, we simulate three canonical scenarios:

1. **Archetype 1: Confident Prior vs. Noisy Sensor** ($\sigma_0 = 1.0, \sigma = 5.0$)
   - The prior has $25\times$ more precision than the sensor. The posterior barely moves from the prior mean.
2. **Archetype 2: Flat / Vague Prior vs. Precision Sensor** ($\sigma_0 = 10.0, \sigma = 1.0$)
   - The sensor has $100\times$ more precision than the prior. The data completely overwhelms the prior.
3. **Archetype 3: Equal Precision Tug-of-War** ($\sigma_0 = 2.0, \sigma = 2.0$)
   - Both sources have equal precision. The posterior settles exactly halfway between them with variance strictly smaller than both ($\sigma_{\text{post}} = \frac{2}{\sqrt{2}} \approx 1.41$).

````r
# Part 3: The Three Archetypes Simulation in Base R
par(mfrow = c(1, 3), mar = c(4.5, 4.5, 3.5, 1.5))
x <- seq(160, 205, length.out = 1000)

scenarios <- list(
  list(title = "Scenario 1: Equal Noise (50/50)", mu_0 = 170.0, s_0 = 5.0, y = 195.0, s = 5.0),
  list(title = "Scenario 2: Precise Sensor (Data Wins)", mu_0 = 170.0, s_0 = 5.0, y = 195.0, s = 1.0),
  list(title = "Scenario 3: Strong Prior (Prior Wins)", mu_0 = 170.0, s_0 = 1.0, y = 195.0, s = 5.0)
)

cat("=== Analytical Gaussian Updating Calculations ===\n")
for (sc in scenarios) {
  mu_0 <- sc$mu_0; s_0 <- sc$s_0
  y <- sc$y; s <- sc$s
  
  w_0 <- 1.0 / (s_0^2)
  w_L <- 1.0 / (s^2)
  
  W_prior <- (w_0 / (w_0 + w_L)) * 100
  W_data  <- (w_L / (w_0 + w_L)) * 100
  
  mu_post <- (w_0 * mu_0 + w_L * y) / (w_0 + w_L)
  sigma_post <- sqrt(1.0 / (w_0 + w_L))
  
  cat(sprintf("%s:\n   w_0: %.4f (%.1f%%) | w_L: %.4f (%.1f%%)\n", sc$title, w_0, W_prior, w_L, W_data))
  cat(sprintf("   Posterior: N(mu=%.2f, sigma=%.2f)\n\n", mu_post, sigma_post))
  
  p_prior <- dnorm(x, mean = mu_0, sd = s_0)
  p_lik   <- dnorm(x, mean = y, sd = s)
  p_post  <- dnorm(x, mean = mu_post, sd = sigma_post)
  
  max_y <- max(c(p_prior, p_lik, p_post)) * 1.1
  
  plot(x, p_post, type = "l", col = "#4CAF50", lwd = 3, ylim = c(0, max_y),
       main = sc$title, xlab = "Parameter mu (Height cm)", ylab = "Density", cex.main = 0.95)
  polygon(c(x, rev(x)), c(p_post, rep(0, length(x))), col = adjustcolor("#4CAF50", alpha.f = 0.2), border = NA)
  lines(x, p_prior, col = "#3F51B5", lwd = 2, lty = 2)
  lines(x, p_lik, col = "#E91E63", lwd = 2, lty = 3)
  
  legend("topright", bty = "n", cex = 0.8,
         legend = c(sprintf("Prior (%.0f%%)", W_prior), sprintf("Data (%.0f%%)", W_data), sprintf("Post N(%.1f, %.2f)", mu_post, sigma_post)),
         col = c("#3F51B5", "#E91E63", "#4CAF50"), lty = c(2, 3, 1), lwd = c(2, 2, 3))
}
par(mfrow = c(1, 1))
````

## Part 4: Scaling to $N$ Observations ($n \to \infty$)

When we collect a sample of $n$ observations $y_1, y_2, \dots, y_n$ with sample mean $\bar{y} = \frac{1}{n} \sum y_i$, the joint likelihood exponent sums over all $n$ points:
$$\sum_{i=1}^n -\frac{(y_i - \mu)^2}{2\sigma^2} = -\frac{1}{2} \left( \frac{n}{\sigma^2} \right) (\mu - \bar{y})^2 + \text{const}$$

Notice the data weight is multiplied by $n$:
* **Prior Weight**: $w_0 = \frac{1}{\sigma_0^2}$
* **Collective Data Weight**: $w_L = \mathbf{\frac{n}{\sigma^2}}$

The full updating formula becomes:
$$\mu_{\text{post}} = \frac{\frac{1}{\sigma_0^2}\mu_0 + \frac{n}{\sigma^2}\bar{y}}{\frac{1}{\sigma_0^2} + \frac{n}{\sigma^2}}$$

$$\sigma_{\text{post}}^2 = \frac{1}{\frac{1}{\sigma_0^2} + \frac{n}{\sigma^2}}$$

### The Inevitable Triumph of Evidence (Bernstein-von Mises Theorem)
As $n \to \infty$, the collective data weight $\frac{n}{\sigma^2}$ grows without bound, while the prior weight $\frac{1}{\sigma_0^2}$ is fixed:
$$\lim_{n \to \infty} W_{\text{data}} = \mathbf{100\%}, \qquad \lim_{n \to \infty} \mu_{\text{post}} = \mathbf{\bar{y}}$$
With large sample sizes in production (e.g., $n = 10,000$), reasonable priors have virtually zero influence on the final posterior mean!

## Part 5: The Luxury and The Catch — When Conjugacy Breaks Down

Gaussian conjugacy is an intellectual masterpiece: it lets engineers fuse sensor streams in real time using simple arithmetic, completely bypassing numerical integration.

### The Catch
However, this analytical luxury rests on two fragile assumptions:
1. **Known, Fixed Noise ($\sigma$)**: We assumed the measurement variance $\sigma^2$ is perfectly known in advance. In practice, sensor noise is often unknown and must be estimated alongside $\mu$.
2. **Strict Bell Curve Shapes**: We assumed both the prior and the measurement noise follow ideal Gaussian bell curves. If latency data has heavy tails or a sensor suffers an outlier glitch, the Normal likelihood can get dragged violently off target.

The moment $\sigma$ is unknown, or the likelihood is non-Gaussian (such as a heavy-tailed Student-$t$ distribution), **analytical conjugacy breaks down completely**. When that happens, we turn to numerical engines:
* **Chapter 5**: **Grid Approximation** over arbitrary 1D/2D parameter meshes.
* **Chapters 6 & 7**: **Laplace Approximation** using curvature and the Hessian matrix.
* **Chapters 8–10**: **Markov Chain Monte Carlo (MCMC)** for general high-dimensional posteriors.

````r
# Part 5 Demo: When Conjugacy Breaks Down (Outlier Glitch vs. Heavy-Tailed Likelihood)
mu_0 <- 10.0; sigma_0 <- 1.0
sigma <- 0.5
obs <- c(10.1, 9.9, 10.2, 18.5)  # Single outlier glitch at 18.5

# 1. Closed-form Normal-Normal conjugate posterior (fragile to outliers)
w_0 <- 1.0 / (sigma_0^2)
w_L <- length(obs) / (sigma^2)
mu_post_norm <- (w_0 * mu_0 + w_L * mean(obs)) / (w_0 + w_L)
sigma_post_norm <- sqrt(1.0 / (w_0 + w_L))

# 2. Non-conjugate Student-t likelihood (df = 2) on a 1D numerical grid
mu_grid <- seq(7.0, 14.0, length.out = 1000)
d_mu <- mu_grid[2] - mu_grid[1]
log_prior <- dnorm(mu_grid, mu_0, sigma_0, log = TRUE)
log_like_t <- sapply(mu_grid, function(m) sum(dt((obs - m) / sigma, df = 2, log = TRUE)))
log_post_t <- log_prior + log_like_t
post_t <- exp(log_post_t - max(log_post_t))
post_t <- post_t / (sum(post_t) * d_mu)
mu_post_t <- sum(mu_grid * post_t) * d_mu

cat(sprintf("Observations: %s\n", paste(obs, collapse = ", ")))
cat(sprintf("Normal Conjugate Posterior Mean: %.3f (pulled by glitch 18.5!)\n", mu_post_norm))
cat(sprintf("Student-t (df=2) Posterior Mean: %.3f (robust to glitch!)\n", mu_post_t))

par(mfrow = c(1, 1), mar = c(4.5, 4.5, 3.0, 1.0))
d_norm_post <- dnorm(mu_grid, mu_post_norm, sigma_post_norm)
plot(mu_grid, post_t, type = "l", col = "darkcyan", lwd = 2.5,
     ylim = c(0, max(post_t, d_norm_post) * 1.15),
     main = "When Conjugacy Breaks Down: Gaussian Fragility vs. Student-t Robustness",
     xlab = "Parameter mu", ylab = "Posterior Density")
lines(mu_grid, d_norm_post, col = "firebrick", lwd = 2.5)
abline(v = 10.0, col = "steelblue", lty = 2, lwd = 1.5)
legend("topright",
       legend = c(sprintf("Normal Conjugate (mean=%.2f)", mu_post_norm),
                  sprintf("Student-t Grid (mean=%.2f)", mu_post_t),
                  "True cluster (~10.0)"),
       col = c("firebrick", "darkcyan", "steelblue"),
       lty = c(1, 1, 2), lwd = c(2.5, 2.5, 1.5), cex = 0.85, bty = "n")
````

### 🟢 Tier 1 (Conceptual): Why Variance Governs Weighting Rather Than Standard Deviation
> **Question**: Two independent astronomers measure the speed of an incoming asteroid:
> * Astronomer A uses an optical telescope with measurement error $\sigma_A = 10\text{ km/s}$.
> * Astronomer B uses an uncalibrated radio telescope with error $\sigma_B = 20\text{ km/s}$.
> 
> Does Astronomer A's reading receive $2\times$ as much weight as Astronomer B's, or something else? Explain why variance (squared error) determines voting power in Bayes' rule rather than standard deviation.

````r
# ============================================================
# 🟢 Tier 1: Your Turn — Attempt the Problem Here
# ============================================================
sigma_A <- 10.0
sigma_B <- 20.0

# Calculate the relative weight ratio w_A / w_B:
# weight_ratio <- ...

# cat(sprintf("Astronomer A gets %.1fx the weight of Astronomer B\n", weight_ratio))
````

<details>
<summary><b>🔍 Click here to view the solution & explanation</b></summary>

**Exact Answer**: Astronomer A receives **$4\times$ as much weight** as Astronomer B (not $2\times$!).

**Mathematical Reason**:
Weights are inversely proportional to **variance** ($\sigma^2$), which is the square of standard deviation:
$$w_A = \frac{1}{\sigma_A^2} = \frac{1}{10^2} = \frac{1}{100} = 0.01$$
$$w_B = \frac{1}{\sigma_B^2} = \frac{1}{20^2} = \frac{1}{400} = 0.0025$$

$$\text{Weight Ratio} = \frac{w_A}{w_B} = \frac{0.01}{0.0025} = \left( \frac{20}{10} \right)^2 = \mathbf{4}$$

Astronomer A carries **$80\%$** of the combined vote ($0.01 / 0.0125$), while Astronomer B carries only **$20\%$**. In Gaussian likelihoods, doubling your instrument's error margin reduces your credibility by a factor of four.

### 🟡 Tier 2 (Guided Calculation): Autonomous Vehicle Sensor Fusion
> **Scenario**: An autonomous vehicle's navigation system is estimating its distance to a pedestrian crossing the street:
> * **Prior Belief (GPS & Dead Reckoning)**: Distance $\mu_0 = 50.0\text{ m}$ with uncertainty $\sigma_0 = 4.0\text{ m}$.
> * **Sensor 1 (Ultrasonic Sonar)**: Measures $y_1 = 44.0\text{ m}$ with noise $\sigma_1 = 3.0\text{ m}$.
> * **Sensor 2 (LIDAR)**: Measures $y_2 = 42.0\text{ m}$ with noise $\sigma_2 = 0.5\text{ m}$.
> 
> Sequentially update the distance estimate:

### 🔴 Tier 3 (Production Challenge): Proof of Strictly Decreasing Variance
> **Question**: In safety-critical aerospace and medical devices, engineers require guarantees that assimilating sensor data will never *increase* system uncertainty.
> 
> Prove mathematically that under Normal-Normal updating with any non-zero noise $\sigma > 0$, the posterior variance is **strictly smaller** than both the prior variance AND the sensor noise:
> $$\sigma_{\text{post}}^2 < \min(\sigma_0^2, \sigma^2)$$


````r
# ============================================================
# 🔴 Tier 3: Your Turn — Attempt the Proof / Verification Here
# ============================================================
# Verify numerically across 1000 random positive pairs (sigma_0, sigma_L):
# sigmas_0 <- runif(1000, 0.1, 100)
# sigmas_L <- runif(1000, 0.1, 100)
# sigmas_post <- sqrt(1.0 / (1.0/sigmas_0^2 + 1.0/sigmas_L^2))
# all_smaller <- all(sigmas_post < sigmas_0 & sigmas_post < sigmas_L)
# cat(sprintf("Always strictly smaller? %s\n", all_smaller))
````

**Formal Mathematical Proof**:
1. By definition, variance and noise are strictly positive real numbers: $\sigma_0^2 > 0$ and $\sigma^2 > 0$.
2. Consequently, their weights are strictly positive:
   $$w_0 = \frac{1}{\sigma_0^2} > 0 \qquad\text{and}\qquad w_L = \frac{1}{\sigma^2} > 0$$
3. The sum of two positive numbers is strictly greater than either individual summand:
   $$w_0 + w_L > w_0 \qquad\text{and}\qquad w_0 + w_L > w_L$$
4. Inverting both sides of an inequality of positive numbers reverses the inequality:
   $$\frac{1}{w_0 + w_L} < \frac{1}{w_0} = \sigma_0^2$$
   $$\frac{1}{w_0 + w_L} < \frac{1}{w_L} = \sigma^2$$
5. Since $\sigma_{\text{post}}^2 = \frac{1}{w_0 + w_L}$, we have proven:
   $$\sigma_{\text{post}}^2 < \sigma_0^2 \qquad\text{and}\qquad \sigma_{\text{post}}^2 < \sigma^2 \quad \blacksquare$$

**Production Guarantee**: Every valid observation $y$ carries positive information ($w_L > 0$). In Bayesian inference, observing data is mathematically guaranteed to contract the uncertainty bounds of the system.
