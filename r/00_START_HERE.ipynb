## Quick Summary of Computational Paradigms

```
┌─────────────────────────────────────────────────────────────────────────────┐
│ 1. Grid Approximation  -->  Exact 2D lattice (1-3 parameters max)           │
│ 2. Quadratic Approx    -->  Parabolic mode + Hessian (Instantaneous 0.001s) │
│ 3. MCMC (Stan / HMC)   -->  Physics simulation / Random Walk (1000s params) │
│ 4. Exact Conjugate     -->  Closed-form Beta-Binomial update (Real-time CI) │
│ 5. Fading Memory Bayes  -->  Exponential discount factor γ (Half-life & ESS) │
└─────────────────────────────────────────────────────────────────────────────┘
```
